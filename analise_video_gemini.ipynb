# Explorando a análise de videos com o Gemini API


## Inicialização do modelo

Abaixo vamos inicializar a chamada da API passando a secret gerada no site AI Studio

```python
from google import genai
from google.genai import types

client = genai.Client(api_key="PASSE_SUA_SECRET")
```

## Explorando a chamada Multimodal

Função criada para facilitar a chamada da API, onde vamos receber o link do video, prompt e o modelo. E ao final é retornado a resposta da API

```python
def chamada_API_processar_video(prompt,modelo, link):
  response = client.models.generate_content(
    model=modelo,
    contents=types.Content(
        parts=[
            types.Part(
                file_data=types.FileData(file_uri=link)
            ),
            types.Part(text=prompt)
        ]
    )
)
  print(response.text)
```
